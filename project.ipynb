# 1. Data Loading & Initial Inspection

In this stage, we load the Fire and No-Fire datasets and perform
an initial inspection of their structure.

### Steps:
1. Load both datasets.
2. Check the number of rows and columns.
3. Compare the column names of both datasets.
4. Inspect the initial data before cleaning.

```python
# Import pandas for data loading and data manipulation
import pandas as pd

# Load the Fire dataset
fire_df = pd.read_csv("gds_fire_30mm.csv")

# Load the No-Fire dataset
nofire_df = pd.read_csv("gds_nofire_30mm.csv")

# Display the number of rows and columns in each dataset
print("Fire shape:", fire_df.shape)
print("No Fire shape:", nofire_df.shape)

# Display the column names of the Fire dataset
print("\nFire columns:")
print(fire_df.columns.tolist())

# Display the column names of the No-Fire dataset
print("\nNo Fire columns:")
print(nofire_df.columns.tolist())
```

```python
# Display the first 5 rows of the Fire dataset
print("Fire Dataset:")
display(fire_df.head())

# Display the first 5 rows of the No-Fire dataset
print("\nNo Fire Dataset:")
display(nofire_df.head())
```

# 2. Data Preparation & Target Creation

In this stage, we prepare the two datasets for machine learning.

### Steps:
1. Remove the `system:index` identifier because it is not a predictive feature.
2. Create a binary target variable:
   - `1` = Fire
   - `0` = No Fire
3. Combine the Fire and No-Fire datasets into one DataFrame.

```python
# Remove the system:index identifier because it does not represent
# a meaningful environmental or geographical feature.
fire_df = fire_df.drop(columns=["system:index"])
nofire_df = nofire_df.drop(columns=["system:index"])
```

```python
# Assign the target class for Fire cases
# 1 = Fire
fire_df["Target"] = 1

# Assign the target class for No-Fire cases
# 0 = No Fire
nofire_df["Target"] = 0

# Check the number of samples in each class
print("Fire:")
print(fire_df["Target"].value_counts())

print("\nNo Fire:")
print(nofire_df["Target"].value_counts())
```

```python
# Combine Fire and No-Fire datasets into one DataFrame
df = pd.concat([fire_df, nofire_df], ignore_index=True)

# Display the final shape
print("Shape:", df.shape)

# Display the distribution of the target classes
print("\nTarget distribution:")
print(df["Target"].value_counts())
```

# 3. Duplicate Detection & Removal

In this stage, we check for duplicate rows in the combined dataset
and remove them to avoid using repeated observations during model training.

### Steps:
1. Count duplicated rows.
2. Remove duplicate rows.
3. Verify that no duplicates remain.

The `Target` column already represents the class we want to predict:
- `1` = Fire
- `0` = No Fire

Since `labels` contains the same information, keeping both columns is unnecessary

### Steps:
1. Compare `labels` and `Target`.
2. Remove the redundant `labels` column.
3. Keep `Target` as the final target variable.

```python
# Compare the class distribution of the labels column
print("labels:")
print(df["labels"].value_counts())

# Compare the class distribution of the Target column
print("\nTarget:")
print(df["Target"].value_counts())
```

```python
# Remove the redundant labels column because Target already
# contains the information needed for classification.
df = df.drop(columns=["labels"])

# Check the dataset shape and remaining columns
print("Shape:", df.shape)
print(df.columns.tolist())
```

# 5. Data Cleaning & Missing Values

In this stage, we identify and handle invalid and missing values
before building the machine learning models.

### Steps:
1. Detect the `-99999` placeholder values.
2. Convert invalid placeholders to `NaN`.
3. Analyze the amount of missing data.
4. Check for invalid feature values and obvious outliers.
5. Keep valid observations and handle missing values later during preprocessing.

```python
# Count the invalid placeholder value (-99999) in each column
invalid_values = (df == -99999).sum()

# Display only columns that contain invalid values
print(invalid_values[invalid_values > 0])
```

```python
import numpy as np

# Create a copy of the dataset for cleaning
df_clean = df.copy()

# Replace the invalid placeholder value with NaN
df_clean = df_clean.replace(-99999, np.nan)

# Count missing values for each row
missing_per_row = df_clean.isna().sum(axis=1)

# Display the distribution of missing values per row
print("Missing values per row:")
print(missing_per_row.value_counts().sort_index())
```

```python
# Check whether important environmental features contain
# physically invalid values.

checks = {
    "aspect < 0": (df_clean["aspect"] < 0).sum(),
    "aspect > 360": (df_clean["aspect"] > 360).sum(),

    "slope < 0": (df_clean["slope"] < 0).sum(),

    "EVI < -1": (df_clean["evimean"] < -1).sum(),
    "EVI > 1": (df_clean["evimean"] > 1).sum(),

    "NDVI < -1": (df_clean["ndvimean"] < -1).sum(),
    "NDVI > 1": (df_clean["ndvimean"] > 1).sum(),

    "NDWI < -1": (df_clean["ndwimean"] < -1).sum(),
    "NDWI > 1": (df_clean["ndwimean"] > 1).sum(),

    "PREC < 0": (df_clean["precmean"] < 0).sum()
}

print(pd.Series(checks))
```

```python
# Define all EVI-related columns
evi_cols = [
    "evimax",
    "evimean",
    "evimedian",
    "evimin",
    "evisum"
]

# Convert the invalid EVI observation to missing values
# instead of keeping an unrealistic measurement.
df_clean.loc[df_clean["evimean"] > 1, evi_cols] = np.nan

# Verify the updated number of missing values
print(df_clean[evi_cols].isna().sum())
```

# 6. Missingness Analysis & Decision

In this stage, we analyze the missing values after converting the
invalid `-99999` placeholders to `NaN`.

The goal is to understand:
- How much data is missing from each feature group.
- Whether missing values are related to the target class.
- Whether dropping rows with missing values is a safe decision.

Based on this analysis, we will choose an appropriate strategy
for handling missing values during model preprocessing.

```python
# Calculate the percentage of missing values in each column
missing_pct = df_clean.isna().mean() * 100

# Display only columns that contain missing values,
# sorted from highest to lowest percentage
print(
    missing_pct[missing_pct > 0]
    .sort_values(ascending=False)
)
```

```python
# Define the environmental feature groups
feature_groups = {
    "EVI": [
        "evimax", "evimean", "evimedian",
        "evimin", "evisum"
    ],

    "LST": [
        "lstmax", "lstmean", "lstmedian",
        "lstmin", "lstsum"
    ],

    "NDVI": [
        "ndvimax", "ndvimean", "ndvimedian",
        "ndvimin", "ndvisum"
    ],

    "NDWI": [
        "ndwimax", "ndwimean", "ndwimedian",
        "ndwimin", "ndwisum"
    ]
}


# Compare the Fire percentage when a complete feature group
# is missing versus when it is available.
for group, cols in feature_groups.items():

    # True if all features in the group are missing
    missing = df_clean[cols].isna().all(axis=1)

    # Fire percentage among rows with the group missing
    missing_fire_rate = df_clean.loc[
        missing, "Target"
    ].mean() * 100

    # Fire percentage among rows where the group is available
    available_fire_rate = df_clean.loc[
        ~missing, "Target"
    ].mean() * 100

    print(f"\n===== {group} =====")
    print(
        "Missing → Fire %:",
        round(missing_fire_rate, 2)
    )
    print(
        "Available → Fire %:",
        round(available_fire_rate, 2)
    )
```

# 7. Feature Correlation Analysis

In this stage, we analyze the relationships between numerical features
and their relationship with the target variable.

### Goals:
1. Detect highly correlated features.
2. Identify redundant information.
3. Measure the relationship between each feature and Fire occurrence.
4. Use these findings to guide feature selection and dimensionality reduction.

```python
# Calculate the correlation matrix for numerical features
# The .geo column is excluded because it contains text.
corr = df_clean.drop(columns=[".geo"]).corr(numeric_only=True)

# Extract unique feature pairs with high positive correlation
high_corr = (
    corr.where(
        np.triu(np.ones(corr.shape), k=1).astype(bool)
    )
    .stack()
    .sort_values(ascending=False)
)

# Display feature pairs with correlation greater than 0.90
print("Highly Correlated Feature Pairs:")
print(high_corr[high_corr > 0.90])
```

```python
# Calculate the absolute correlation between each numerical feature
# and the target variable.
target_corr = (
    df_clean
    .corr(numeric_only=True)["Target"]
    .abs()
    .sort_values(ascending=False)
)

# Remove the Target itself from the results
target_corr = target_corr.drop("Target")

# Display the most related features
print("Feature correlation with Target:")
print(target_corr)
```

# 8. Feature Selection — Removing LST Features

In this stage, we reduce the feature set by removing the LST
(Land Surface Temperature) features.

LST has the highest missing-value percentage in our dataset
(about 61.39%), which makes these features less reliable for modeling.

### Decision:
- Remove: `lstmax`, `lstmean`, `lstmedian`, `lstmin`, `lstsum`
- Keep the remaining environmental, geographical, and temporal features.

This helps simplify the dataset and reduces the impact of excessive missing data.

```python
# List all LST-related features
lst_cols = [
    "lstmax",
    "lstmean",
    "lstmedian",
    "lstmin",
    "lstsum"
]

# Remove the LST features from the cleaned dataset
df_model = df_clean.drop(columns=lst_cols)

# Display the new dataset shape
print("Shape:", df_model.shape)

# Display the remaining features
print(df_model.columns.tolist())
```

# 9. Geographical Feature Engineering

In this stage, we convert the `.geo` column into useful numerical
geographical features.

The `.geo` column contains location information as JSON text, which
cannot be directly used effectively by the machine learning model.

### Steps:
1. Extract latitude from `.geo`.
2. Extract longitude from `.geo`.
3. Store them as numerical features.
4. Remove the original `.geo` column.

```python
import json

# Extract latitude and longitude from the JSON geometry column
def extract_coordinates(value):
    try:
        # Convert the JSON string into a Python dictionary
        data = json.loads(value)

        # Get longitude and latitude from the coordinates
        longitude, latitude = data["coordinates"]

        # Return latitude first, then longitude
        return pd.Series([latitude, longitude])

    except:
        # Return missing values if the geometry cannot be parsed
        return pd.Series([np.nan, np.nan])


# Create two new numerical features from the .geo column
df_model[["latitude", "longitude"]] = df_model[".geo"].apply(
    extract_coordinates
)

# Check the extracted coordinates
print(df_model[[".geo", "latitude", "longitude"]].head())
```

```python
# Remove the original geometry column because
# the useful location information is now represented
# by latitude and longitude.
df_model = df_model.drop(columns=[".geo"])

# Check the new dataset shape
print("Shape:", df_model.shape)

# Display the remaining features
print(df_model.columns.tolist())
```

# 10. Time Feature Engineering

In this stage, we convert the `time` feature into meaningful
temporal features.

The original `time` value is stored as an integer in the format
`YYYYMMDD`, so it is converted into a date first.

### Steps:
1. Convert `time` into a datetime value.
2. Extract `year`, `month`, and `day_of_year`.
3. Remove the original time representation and other temporary
   date-related features that are not used by the final model.

```python
# Convert the time feature from YYYYMMDD integer format
# into a proper datetime feature.
df_model["date"] = pd.to_datetime(
    df_model["time"].astype(str),
    format="%Y%m%d"
)

# Extract useful temporal features
df_model["year"] = df_model["date"].dt.year
df_model["month"] = df_model["date"].dt.month
df_model["day_of_month"] = df_model["date"].dt.day
df_model["day_of_year"] = df_model["date"].dt.dayofyear

# Display the generated time features
print(
    df_model[
        ["time", "date", "year", "month",
         "day_of_month", "day_of_year"]
    ].head()
)
```

```python
# Define time-related columns that will not be used
# directly by the final machine learning model.
time_drop_cols = [
    "time",
    "date",
    "day_of_month"
]

# Remove redundant time representations
df_model_v2 = df_model.drop(columns=time_drop_cols)

# Display the final shape and feature list
print("Shape:", df_model_v2.shape)
print(df_model_v2.columns.tolist())
```

# 11. Final Dataset Inspection & Train/Test Split

In this stage, we prepare the final dataset for machine learning.

### Steps:
1. Separate the input features (`X`) from the target variable (`y`).
2. Check the final feature set.
3. Split the data into training and testing sets.
4. Use stratification to preserve the Fire/No Fire class distribution.

```python
# Separate the input features from the target variable
X = df_model_v2.drop(columns=["Target"])
y = df_model_v2["Target"]

# Display the final feature count and dataset size
print("Features:", X.shape)
print("Target:", y.shape)

# Display the final feature names
print("\nFeature names:")
print(X.columns.tolist())
```

```python
from sklearn.model_selection import train_test_split

# Split the dataset into training and testing sets
# 80% for training and 20% for testing.
# stratify=y keeps the same Fire/No Fire ratio in both sets.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display the shapes of the resulting datasets
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
# Check the class distribution in the training set
print("Training target distribution:")
print(y_train.value_counts())

# Check the class distribution in the testing set
print("\nTesting target distribution:")
print(y_test.value_counts())
```

# 12. Data Preprocessing

In this stage, we prepare the features for machine learning.

### Steps:
1. Handle missing values using median imputation.
2. Standardize numerical features using StandardScaler.
3. Handle class imbalance using SMOTE.

These preprocessing steps will later be placed inside the
machine learning pipeline to prevent data leakage during
cross-validation.

```python
# Import preprocessing tools
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import SMOTE

# Create a median imputer
# The median is calculated from the training data only.
imputer = SimpleImputer(strategy="median")

# Fit the imputer on the training data and transform it
X_train_imputed = imputer.fit_transform(X_train)

# Apply the same imputer to the test data
X_test_imputed = imputer.transform(X_test)

# Check the resulting shapes
print("X_train after imputation:", X_train_imputed.shape)
print("X_test after imputation:", X_test_imputed.shape)
```

```python
# Create a StandardScaler
scaler = StandardScaler()

# Fit the scaler on the training data and transform it
X_train_scaled = scaler.fit_transform(X_train_imputed)

# Apply the same scaling parameters to the test data
X_test_scaled = scaler.transform(X_test_imputed)

# Check the resulting shapes
print("X_train after scaling:", X_train_scaled.shape)
print("X_test after scaling:", X_test_scaled.shape)
```

```python
# Create SMOTE to balance the training classes
smote = SMOTE(random_state=42)

# Generate synthetic samples for the minority class
# SMOTE is applied only to the training data.
X_train_smote, y_train_smote = smote.fit_resample(
    X_train_scaled,
    y_train
)

# Display class distribution before and after SMOTE
print("Before SMOTE:")
print(y_train.value_counts())

print("\nAfter SMOTE:")
print(y_train_smote.value_counts())

print("\nTraining shape before:", X_train_scaled.shape)
print("Training shape after :", X_train_smote.shape)
```

# 13. Baseline Decision Tree Model

In this stage, we build a baseline Decision Tree classifier.

The purpose of the baseline model is to establish an initial
performance level before applying hyperparameter tuning.

### Steps:
1. Create a Decision Tree classifier.
2. Train the model on the preprocessed training data.
3. Predict the Fire/No Fire classes on the test set.
4. Evaluate the model using classification metrics.

```python
# Import the Decision Tree classifier
from sklearn.tree import DecisionTreeClassifier

# Create the baseline Decision Tree model
# random_state ensures reproducible results.
dt_model = DecisionTreeClassifier(
    random_state=42
)

# Train the model on the balanced training data
dt_model.fit(
    X_train_smote,
    y_train_smote
)
```

```python
# Generate predictions on the test data
y_pred_dt = dt_model.predict(X_test_scaled)

# Import evaluation metrics
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Calculate the main evaluation metrics
accuracy = accuracy_score(y_test, y_pred_dt)
precision = precision_score(y_test, y_pred_dt)
recall = recall_score(y_test, y_pred_dt)
f1 = f1_score(y_test, y_pred_dt)

# Display the results
print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))

# Display the detailed classification report
print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_dt,
        target_names=["No Fire", "Fire"]
    )
)

# Display the confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_dt))
```

# 14. Decision Tree Hyperparameter Tuning

In this stage, we optimize the Decision Tree hyperparameters
to improve its performance.

The main parameters we tune are:
- `max_depth`
- `min_samples_split`
- `min_samples_leaf`

We use GridSearchCV with cross-validation and F1-score as the
scoring metric because detecting Fire cases is important.

The final pipeline also includes preprocessing and SMOTE inside
the cross-validation process to avoid data leakage.

```python
# Import the required tools
from imblearn.pipeline import Pipeline
from imblearn.over_sampling import SMOTE
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV

# Build a complete Decision Tree pipeline
# Each preprocessing step is fitted only on the training portion
# of each cross-validation fold.
dt_pipeline = Pipeline([
    
    # Handle missing values using the median
    ("imputer", SimpleImputer(strategy="median")),
    
    # Standardize the numerical features
    ("scaler", StandardScaler()),
    
    # Balance the classes using SMOTE only on training folds
    ("smote", SMOTE(random_state=42)),
    
    # Decision Tree classifier
    ("classifier", DecisionTreeClassifier(random_state=42))
])
```

```python
# Define the hyperparameters to search
param_grid_dt = {
    "classifier__max_depth": [5, 10, 15, 20],
    "classifier__min_samples_split": [2, 5, 10],
    "classifier__min_samples_leaf": [1, 2, 5]
}

# Search all parameter combinations using 3-fold cross-validation
# F1-score is used to balance Precision and Recall for Fire detection.
grid_search_dt = GridSearchCV(
    estimator=dt_pipeline,
    param_grid=param_grid_dt,
    cv=3,
    scoring="f1",
    n_jobs=-1
)

# Train the pipeline using the original training data.
# Imputation, scaling, and SMOTE are handled safely inside the pipeline.
grid_search_dt.fit(X_train, y_train)

# Display the best parameters
print("Best Parameters:")
print(grid_search_dt.best_params_)

# Display the best cross-validation F1-score
print("\nBest CV F1-Score:")
print(grid_search_dt.best_score_)
```

```python
# Use the best tuned Decision Tree pipeline to predict
# the unseen test data.
y_pred_best_dt = grid_search_dt.predict(X_test)

# Evaluate the tuned model
print("Classification Report:")
print(
    classification_report(
        y_test,
        y_pred_best_dt,
        target_names=["No Fire", "Fire"]
    )
)

# Display the confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_best_dt))
```

# 15. Support Vector Machine (SVM)

In this stage, we build and tune a Support Vector Machine classifier.

SVM is useful for classification problems where the classes may not
be separated by a simple linear boundary.

We use:
- RBF kernel for non-linear classification.
- StandardScaler because SVM is sensitive to feature scale.
- RandomizedSearchCV to search for good hyperparameters.
- F1-score as the optimization metric.

```python
# Import tools required for SVM
from sklearn.model_selection import RandomizedSearchCV
from imblearn.pipeline import Pipeline
from imblearn.over_sampling import SMOTE
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

# Build the complete SVM pipeline
# Keeping preprocessing and SMOTE inside the pipeline
# prevents data leakage during cross-validation.
svm_pipeline = Pipeline([

    # Handle missing values using the median
    ("imputer", SimpleImputer(strategy="median")),

    # Standardize features because SVM is scale-sensitive
    ("scaler", StandardScaler()),

    # Balance the classes only inside training folds
    ("smote", SMOTE(random_state=42)),

    # SVM classifier
    ("classifier", SVC())
])
```

```python
# Define the SVM hyperparameter search space
param_distributions_svm = {
    "classifier__C": [0.1, 1, 10, 100],
    "classifier__gamma": ["scale", "auto"],
    "classifier__kernel": ["rbf", "linear"]
}

# RandomizedSearchCV tests a limited number of combinations,
# making the search faster than testing every combination.
random_search_svm = RandomizedSearchCV(
    estimator=svm_pipeline,
    param_distributions=param_distributions_svm,
    n_iter=8,
    cv=3,
    scoring="f1",
    random_state=42,
    n_jobs=-1
)

# Fit the search using the original training data.
# Imputation, scaling, and SMOTE are handled inside the pipeline.
random_search_svm.fit(X_train, y_train)

# Display the best hyperparameters
print("Best SVM Parameters:")
print(random_search_svm.best_params_)

# Display the best cross-validation F1-score
print("\nBest SVM CV F1-Score:")
print(random_search_svm.best_score_)
```

```python
# Predict Fire/No Fire on the unseen test set
y_pred_svm = random_search_svm.predict(X_test)

# Evaluate the best SVM model
print("Classification Report:")
print(
    classification_report(
        y_test,
        y_pred_svm,
        target_names=["No Fire", "Fire"]
    )
)

# Display the confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_svm))
```

# 16. XGBoost Model

In this stage, we build and tune an XGBoost classifier.

XGBoost is a gradient boosting algorithm that combines multiple
decision trees to build a stronger predictive model.

### Steps:
1. Build an XGBoost pipeline.
2. Handle missing values using median imputation.
3. Handle class imbalance using SMOTE.
4. Tune the XGBoost hyperparameters using RandomizedSearchCV.
5. Evaluate the best model on the test set.

```python
# Import the required tools for XGBoost
from xgboost import XGBClassifier
from imblearn.pipeline import Pipeline
from imblearn.over_sampling import SMOTE
from sklearn.impute import SimpleImputer
from sklearn.model_selection import RandomizedSearchCV

# Build the XGBoost pipeline
# XGBoost does not require feature scaling.
xgb_pipeline = Pipeline([

    # Handle missing values using the median
    ("imputer", SimpleImputer(strategy="median")),

    # Balance the Fire and No Fire classes
    # SMOTE is applied only to the training portion of each fold
    ("smote", SMOTE(random_state=42)),

    # XGBoost classifier
    ("classifier", XGBClassifier(
        random_state=42,
        eval_metric="logloss"
    ))
])
```

```python
# Define the XGBoost hyperparameter search space
param_distributions_xgb = {
    "classifier__n_estimators": [100, 200, 300, 400],
    "classifier__max_depth": [3, 4, 5, 6, 8],
    "classifier__learning_rate": [0.01, 0.03, 0.05, 0.1],
    "classifier__subsample": [0.7, 0.8, 0.9, 1.0],
    "classifier__colsample_bytree": [0.7, 0.8, 0.9, 1.0],
    "classifier__min_child_weight": [1, 3, 5]
}

# Use RandomizedSearchCV to reduce training time
# 10 parameter combinations × 3 cross-validation folds
random_search_xgb = RandomizedSearchCV(
    estimator=xgb_pipeline,
    param_distributions=param_distributions_xgb,
    n_iter=10,
    cv=3,
    scoring="f1",
    random_state=42,
    n_jobs=-1
)

# Train the search using the original training data
# All preprocessing steps are performed safely inside the pipeline.
random_search_xgb.fit(X_train, y_train)

# Display the best hyperparameters
print("Best XGBoost Parameters:")
print(random_search_xgb.best_params_)

# Display the best cross-validation F1-score
print("\nBest XGBoost CV F1-Score:")
print(random_search_xgb.best_score_)
```

```python
# Use the best XGBoost model to predict the unseen test set
y_pred_xgb = random_search_xgb.predict(X_test)

# Evaluate the tuned XGBoost model
print("Classification Report:")
print(
    classification_report(
        y_test,
        y_pred_xgb,
        target_names=["No Fire", "Fire"]
    )
)

# Display the confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_xgb))
```

# 17. Model Comparison & Selection

In this stage, we compare the trained machine learning models.

The main evaluation metrics are:
- Accuracy
- Precision
- Recall
- F1-Score

Since the main objective is detecting Fire cases, F1-Score and
Recall for the Fire class are particularly important.

# 20. ROC-AUC and Probability Threshold Selection

In this stage, we evaluate the probability output of the final
XGBoost model and select a decision threshold.

### Goals:
1. Calculate Fire probabilities.
2. Evaluate the model using ROC-AUC.
3. Use the validation set to select an appropriate classification threshold.
4. Apply the selected threshold to the test predictions.

The final threshold selected for the application is 0.75.

```python
# Import ROC-AUC evaluation metric
from sklearn.metrics import roc_auc_score

# Get the predicted probability of Fire (class 1)
y_prob_top10 = xgb_top10_pipeline.predict_proba(
    X_test_top10
)[:, 1]

# Calculate ROC-AUC using the Fire probabilities
roc_auc_top10 = roc_auc_score(
    y_test,
    y_prob_top10
)

print("ROC-AUC:", round(roc_auc_top10, 4))
```

```python
from sklearn.model_selection import train_test_split

# Split the training data into training and validation sets
X_train_fit, X_val, y_train_fit, y_val = train_test_split(
    X_train_top10,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

print("Training samples:", X_train_fit.shape[0])
print("Validation samples:", X_val.shape[0])
```

```python
# Train the XGBoost model on the training portion only
xgb_top10_threshold_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("smote", SMOTE(random_state=42)),
    ("classifier", XGBClassifier(
        n_estimators=400,
        max_depth=6,
        min_child_weight=3,
        learning_rate=0.1,
        subsample=0.7,
        colsample_bytree=1.0,
        random_state=42,
        eval_metric="logloss"
    ))
])

xgb_top10_threshold_model.fit(
    X_train_fit,
    y_train_fit
)
```

```python
# Build the temporary XGBoost model for threshold selection
threshold_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("smote", SMOTE(random_state=42)),
    ("classifier", XGBClassifier(
        n_estimators=400,
        max_depth=6,
        min_child_weight=3,
        learning_rate=0.1,
        subsample=0.7,
        colsample_bytree=1.0,
        random_state=42,
        eval_metric="logloss"
    ))
])

# Train the model on the training portion
threshold_model.fit(
    X_train_fit,
    y_train_fit
)
```

```python
# Calculate Fire probabilities on the validation set
y_prob_val = threshold_model.predict_proba(
    X_val
)[:, 1]

print("Validation probabilities calculated.")
```

```python
from sklearn.metrics import fbeta_score

# Try different probability thresholds
thresholds = np.arange(0.10, 0.95, 0.05)

best_threshold = 0
best_f2 = 0

for threshold in thresholds:

    y_pred_val = (
        y_prob_val >= threshold
    ).astype(int)

    f2 = fbeta_score(
        y_val,
        y_pred_val,
        beta=2
    )

    if f2 > best_f2:
        best_f2 = f2
        best_threshold = threshold

print("Best Threshold:", round(best_threshold, 2))
print("Best Validation F2-Score:", round(best_f2, 4))
```

```python
# Store the threshold selected using validation recall
final_threshold = best_threshold

print("Final Threshold:", final_threshold)
```

```python
# Convert test probabilities into binary predictions
y_pred_threshold = (
    y_prob_top10 >= final_threshold
).astype(int)

print("Top-10 XGBoost with Threshold =", final_threshold)
print("------------------------------------------")

print(
    f"Accuracy : {accuracy_score(y_test, y_pred_threshold):.4f}"
)

print(
    f"Precision: {precision_score(y_test, y_pred_threshold):.4f}"
)

print(
    f"Recall   : {recall_score(y_test, y_pred_threshold):.4f}"
)

print(
    f"F1-Score : {f1_score(y_test, y_pred_threshold):.4f}"
)
```
